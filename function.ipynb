## Default argument function

- if provide any value, python use it
- otherwise, default value using

```python
def name(name = "Gurst"):
    print("Hello,",name)

name()
```

```python
name("Naim")
```

### Explanation:

- In the first call greet(), no argument is passed, so Python uses the default value "Guest".
- In the second call greet("Kate"), the default value is overridden with "Naim".

## Example 2:

calling a function using keyword arguments.

```python
citis = ["Delhi", "Gurgoun","Nodia", "Pune", "mumbai", "chennai"]

heros = ["thor", "ironman", "captain America", "shaktiman"]


print(heros[0], end=" ")
print(heros[1], end=" ")

def print_len(list):
    print(len(list))



print_len(citis)
print_len(heros)
```

```python
def print_len(list):
    for item in list:
        print(item, end = " ")



print_len(heros)
print()
```
