## Default argument function

- if provide any value, python use it
- otherwise, default value using

```python
def name(name = "Gurst"):
    print("Hello,",name)

name()
```

```python
name("Naim")
```

- In the first call greet(), no argument is passed, so Python uses the default value "Guest".
- In the second call greet("Kate"), the default value is overridden with "Naim".

## Calcukate Factorial.

```python
n= 5
fact = 1

for i in range (1, n+1):
    fact *= i
print(fact)
```

```python
# In function

def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact = fact*i
    print(fact)

factorial(5)
```

```python
def converter(usd):
    bd = usd * 120
    print(usd, "USD  = ", bd, " Taka")

converter(1)
```

```python
converter(100)
```

```python
def OddEven(n):
    if n % 2 == 1:
        print(n," is Odd")
    else:
        print(n," is Even")

OddEven(9)
OddEven(16)
```

## Recursion

```python
def show(n):
    print(n)

show(5)
```

```python
def show(n):
    if(n == 0):
        return
    print(n)
    show(n-1)
```

```python
show(10)
```

```python
def show(n):
    if(n == 10):
        return
    print(n)
    show(n+1)
show(1)
```
