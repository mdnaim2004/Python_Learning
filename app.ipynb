## Q. Why do we need the print() function?

```python
# Why we use print() in Python

# 1. Show output
print("Hello")

# 2. See results
x = 5 + 3
print(x)  # 8

# 3. Debugging
print("Checking value:", x)

# 4. User interaction
# name = input("Enter name: ")
name = "Naim"
print("Hi", name)
```

At its core, print() is how your program communicates with you.

When Python runs code, it does all the calculations and logic silently in the background. Without print(), you wouldn’t see anything happen

## Why use print():

- print() lets you see what your program is doing
- It displays output on the screen
- Helps verify results of calculations
- Useful for debugging errors
- Allows simple interaction with users
- Can print text, numbers, and variables together
