```python
num = [34, 45, 56, 57, 78, 23, 67, 89, 60, 90, 80, 55, 35]

odds = []

for num in num:
    if num % 2 == 1 and num % 5 == 0:
        odds.append(num)

print(odds)

odd_nums = [n for n in odds]
print(odd_nums)
```

```python
odd_nums = [n for n in odds if num % 2 == 1]
print(odd_nums)
```

```python
players = ['sakib', 'musfik', 'liton']

ages = [38, 37, 32]
age_comb = []

for player in players:
    print('player:', player)
    for age in ages:
        print(player, age)
        age_comb.append([player, age])
print(age_comb)

print("\n")

age_comb2 = [(player, age) for player in players]
print(age_comb2)
```
